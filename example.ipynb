## **<u>calculating with truss_fem_2d</u>**

```python
from src.truss_fem_2d import *

# CREATE STATIC SYSTEM
SYSTEM = StaticSystem()

# ADD NODES
SYSTEM.add_node(x=0.00, y=0.00, x_dof=DegreeOfFreedom.FIXED, y_dof=DegreeOfFreedom.FIXED)
SYSTEM.add_node(x=0.90, y=0.00, y_dof=DegreeOfFreedom.FIXED)
SYSTEM.add_node(x=0.70, y=0.40)
SYSTEM.add_node(x=1.30, y=0.40, Fx=5, Fy=-12)

# ADD ELEMENTS
SYSTEM.add_element(start_node=1, end_node=2)
SYSTEM.add_element(start_node=1, end_node=3)
SYSTEM.add_element(start_node=3, end_node=2)
SYSTEM.add_element(start_node=2, end_node=4)
SYSTEM.add_element(start_node=3, end_node=4)

# PRINT NORMAL FORCES
RESULTS = LinearTrussSolver(SYSTEM)
RESULTS.print_results()
```

## **<u>calculating by hand</u>**
### **<u>support forces</u>**
$\sum H = 0$  
$\,\,\,\,\,$ $\rightarrow 0 = A_x + F_x$  
$\,\,\,\,\,$ $\rightarrow A_x \approx -5,00 \,[kN]$  

$\sum M^{A} = 0$  
$\,\,\,\,\,$ $\rightarrow 0 = F_y \cdot 1,30 -F_x \cdot 0,40 + B_y \cdot 0,90$  
$\,\,\,\,\,$ $\rightarrow 0 = (-12) \cdot 1,30 -5 \cdot 0,40 + B_y \cdot 0,90$  
$\,\,\,\,\,$ $\rightarrow B_y = \frac{176}{9} \approx 19,56 \,[kN]$  

$\sum V = 0$  
$\,\,\,\,\,$ $\rightarrow 0 = A_y + B_y + F_y$  
$\,\,\,\,\,$ $\rightarrow 0 = A_y + \frac{176}{9} + (-12)$  
$\,\,\,\,\,$ $\rightarrow B_y = -\frac{68}{9} \approx -7,56 \,[kN]$  


### **<u>normal forces</u>**  
**<u>node [4]</u>**  
$\sum F_y = 0$  
$\,\,\,\,\,$ $\rightarrow 0 = F_y - N_4 \cdot \frac{0,40}{\sqrt{0,40^2+0,40^2}}$  
$\,\,\,\,\,$ $\rightarrow 0 = -12 - N_4 \cdot \frac{0,40}{\sqrt{0,40^2+0,40^2}}$  
$\,\,\,\,\,$ $\rightarrow N_4 \approx -16,97 \,[kN]$  

$\sum F_x = 0$  
$\,\,\,\,\,$ $\rightarrow 0 = F_x - N_4 \cdot \frac{0,40}{\sqrt{0,40^2+0,40^2}} - N_5$  
$\,\,\,\,\,$ $\rightarrow 0 = 5 - (-16,97) \cdot \frac{0,40}{\sqrt{0,40^2+0,40^2}} - N_5$  
$\,\,\,\,\,$ $\rightarrow N_5 \approx 17,00 \,[kN]$  


**<u>node [2]</u>**  
$\sum F_y = 0$  
$\,\,\,\,\,$ $\rightarrow 0 = B_y + N_3 \cdot \frac{0,40}{\sqrt{0,40^2 + 0,20^2}} + N_4 \cdot \frac{0,40}{\sqrt{0,40^2 + 0,40^2}}$  
$\,\,\,\,\,$ $\rightarrow 0 = 19,56 + N_3 \cdot \frac{0,40}{\sqrt{0,40^2 + 0,20^2}} + (-16,97) \cdot \frac{0,40}{\sqrt{0,40^2 + 0,40^2}}$  
$\,\,\,\,\,$ $\rightarrow N_3 \approx -8,45 \,[kN]$  

$\sum F_x = 0$  
$\,\,\,\,\,$ $\rightarrow 0 = -N_1 - N_3 \cdot \frac{0,20}{\sqrt{0,20^2 + 0,40^2}} + N_4 \cdot \frac{0,40}{\sqrt{0,40^2 + 0,40^2}}$  
$\,\,\,\,\,$ $\rightarrow 0 = -N_1 - (-8,45) \cdot \frac{0,20}{\sqrt{0,20^2 + 0,40^2}} + (-16,97) \cdot \frac{0,40}{\sqrt{0,40^2 + 0,40^2}}$  
$\,\,\,\,\,$ $\rightarrow N_1 \approx -8,22 \,[kN]$  


**<u>node [3]</u>**  
$\sum F_y = 0$  
$\,\,\,\,\,$ $\rightarrow 0 = -N_2 \cdot \frac{0,40}{\sqrt{0,70^2+0,40^2}} -N_3 \cdot \frac{0,40}{\sqrt{0,20^2+0,40^2}}$  
$\,\,\,\,\,$ $\rightarrow 0 = -N_2 \cdot \frac{0,40}{\sqrt{0,70^2+0,40^2}} -(-8,45) \cdot \frac{0,40}{\sqrt{0,20^2+0,40^2}}$  
$\,\,\,\,\,$ $\rightarrow N_2 \approx 15,23 \,[kN]$
