## **<u>calculating by hand</u>**
### **<u>support forces</u>**
$\sum M^{B} = 0$  
$\,\,\,\,\,$ $\rightarrow 0 = -A_y \cdot 0,70 - 12 \cdot 0,40 - 5 \cdot 0,40$  
$\,\,\,\,\,$ $\rightarrow A_y \approx -9,71 \,[kN]$

$\sum V = 0$  
$\,\,\,\,\,$ $\rightarrow 0 = -9,71 + (-12) + B_y$  
$\,\,\,\,\,$ $\rightarrow B_y \approx 24,71 \,[kN]$  

$\sum H = 0$  
$\,\,\,\,\,$ $\rightarrow 0 = A_x + 5$  
$\,\,\,\,\,$ $\rightarrow x \approx -5,00 \,[kN]$  


### **<u>normal forces</u>**
#### **node [4]**
$\sum F_V = 0$  
$\,\,\,\,\,$ $\rightarrow 0 = F_y - N_4 \cdot \frac{0,40}{\sqrt{0,40^2+0,40^2}}$  
$\,\,\,\,\,$ $\rightarrow 0 = -12 - N_4 \cdot \frac{0,40}{\sqrt{0,40^2+0,40^2}}$  
$\,\,\,\,\,$ $\rightarrow N_4 \approx -16,97 \,[kN]$  

$\sum F_H = 0$  
$\,\,\,\,\,$ $\rightarrow 0 = F_x - N_4 \cdot \frac{0,40}{\sqrt{0,40^2+0,40^2}} - N_5$  
$\,\,\,\,\,$ $\rightarrow 0 = 5 - N_4 \cdot \frac{0,40}{\sqrt{0,40^2+0,40^2}} - N_5$  
$\,\,\,\,\,$ $\rightarrow N_5 \approx 17,00 \,[kN]$  


#### **node [2]**
$\sum F_V = 0$  
$\,\,\,\,\,$ $\rightarrow 0 = B_y + N_3 \cdot \frac{0,40}{\sqrt{0,40^2 + 0,20^2}} + N_4 \cdot \frac{0,40}{\sqrt{0,40^2 + 0,40^2}}$  
$\,\,\,\,\,$ $\rightarrow 0 = 24,71 + N_3 \cdot \frac{0,40}{\sqrt{0,40^2 + 0,20^2}} + (-16,97) \cdot \frac{0,40}{\sqrt{0,40^2 + 0,40^2}}$  
$\,\,\,\,\,$ $\rightarrow N_3 \approx -14,21 \,[kN]$  

$\sum F_H = 0$  
$\,\,\,\,\,$ $\rightarrow 0 = -N_1 - N_3 \cdot \frac{0,20}{\sqrt{0,20^2 + 0,40^2}} + N_4 \cdot \frac{0,40}{\sqrt{0,40^2 + 0,40^2}}$  
$\,\,\,\,\,$ $\rightarrow 0 = -N_1 - (-14,21) \cdot \frac{0,20}{\sqrt{0,20^2 + 0,40^2}} + (-16,97) \cdot \frac{0,40}{\sqrt{0,40^2 + 0,40^2}}$  
$\,\,\,\,\,$ $\rightarrow N_3 \approx -5,64 \,[kN]$  

#### **node [3]**
$\sum F_H = 0$  
$\,\,\,\,\,$ $\rightarrow 0 = N_5 + N_3 \cdot \frac{0,20}{\sqrt{0,20^2 + 0,40^2}} - N_2 \cdot \frac{0,70}{\sqrt{0,70^2 + 0,40^2}}$  
$\,\,\,\,\,$ $\rightarrow 0 = 17,00 + (-14,21) \cdot \frac{0,20}{\sqrt{0,20^2 + 0,40^2}} - N_2 \cdot \frac{0,70}{\sqrt{0,70^2 + 0,40^2}}$  
$\,\,\,\,\,$ $\rightarrow N_2 \approx 12,26 \,[kN]$
